```python
import numpy as np
import matplotlib.pyplot as plt

img = np.array([
    [[255, 0, 0],   [0, 255, 0],   [0, 0, 255],   [255, 255, 255]],
    [[255, 255, 0], [0, 255, 255], [255, 0, 255], [0, 0, 0]],
    [[128, 128, 128], [255, 128, 0], [128, 0, 255], [100, 150, 200]]
], dtype=np.uint8)

print(img.shape)
```

```python
plt.imshow(img, interpolation="nearest")
plt.xticks(range(4))
plt.yticks(range(3))
plt.show()
```

```python
# accessing pixel values
img[0, 1]
```

```python
# accessing an enitre row
img[0] 
# same as img[0, :, :]
```

```python
# accessing an entire column
img[:, 0, :]
```

```python
# Access an entire colour channel
img[:, :, 0]  # Red Channel 
```

```python
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]
```

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 4))

axes[0].imshow(img)
axes[0].set_title("RGB")

axes[1].imshow(red, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Red channel")

axes[2].imshow(green, cmap="gray", vmin=0, vmax=255)
axes[2].set_title("Green channel")

axes[3].imshow(blue, cmap="gray", vmin=0, vmax=255)
axes[3].set_title("Blue channel")

plt.show()
```

```python
flipped = img[:, ::-1, :]

plt.imshow(flipped)
plt.show()
```

```python
flipped = img[::-1, :, :]

plt.imshow(flipped)
plt.show()
```
